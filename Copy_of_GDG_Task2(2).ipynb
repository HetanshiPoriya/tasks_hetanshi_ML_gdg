Now, let's some employ some Feature Engineering tactics which you would have performed in the previous tasks.

Create a function called crab_processing which takes a dataframe as an input, performs the following tasks, and returns a cleaned and edited df:

1) Drop the 'id' column.

2) Drop all rows where 'Height' of the crab is 0.

3) Create a new column called 'LostWeight' which is calculated using the following formula - ```
Weight - (Shucked + Viscera + Shell)``` .
If this formula gives a number less than 0, then assign 0 to the LostWeight column, else assign 1.

4) Finally, use one hot encoding to encode the column 'Sex'

Take help from your previous task where you have implemented these steps in different cells.


````python
# Your code ends here
import pandas as pd
import numpy as np

def crab_processing(df: pd.DataFrame) -> pd.DataFrame:
    df = df.drop('id', axis=1)
    df = df[df['Height'] != 0].copy()
    internal_weight_sum = df['Shucked Weight'] + df['Viscera Weight'] + df['Shell Weight']
    raw_lost_weight = df['Weight'] - internal_weight_sum
    df['LostWeight'] = np.where(raw_lost_weight < 0, 0, 1)
    df = pd.get_dummies(df, columns=['Sex'], drop_first=False)

    # Return the cleaned and edited DataFrame
    return df

# Assuming 'processed_df' holds your final data
processed_df = crab_processing(crab_df.copy())
print(processed_df.head())
df_crab = processed_df
````

### Choosing optimal 'k'

**Above, we chose n_neighbours to be equal to 1. Choosing a small value of K leads to unstable decision boundaries. <br>
We need to select n_neighbours by calculating the accuracy for every value of n from 1 to 60 and then choosing the one which gives the highest accuracy.**

````python
# Your code starts here.
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
import numpy as np
accuracy_scores = []
k_values = range(1, 61)

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    k_pred = knn.predict(X_test)
    acc = accuracy_score(y_test, k_pred)
    accuracy_scores.append(acc)
print("Accuracy calculation complete.")
# Your code ends here
````

**Plot a graph of K value vs Accuracy**

````python
# Your code starts here.
import matplotlib.pyplot as plt

plt.figure(figsize=(12, 6))
plt.plot(k_values, accuracy_scores, marker='o', linestyle='--', color='blue', markersize=4)
plt.title('Accuracy vs. K Value for KNN Classifier')
plt.xlabel('K Value (Number of Neighbors)')
plt.ylabel('Accuracy Score')
plt.xticks(np.arange(1, 61, 5)) # Show x-ticks every 5 units for clarity
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
# Your code ends here
````

**Choose the best value of n_neighbours and give a reason why and also print the accuracy**

````python
# Your code starts here.
best_k_index = np.argmax(accuracy_scores)
best_k = k_values[best_k_index]
best_accuracy = accuracy_scores[best_k_index]

print(f"The best K value (n_neighbors) is: {best_k}")
print(f"The highest accuracy achieved is: {best_accuracy:.4f}")
# Your code ends here
````

````python
# Reasons:
The value K=5 is chosen because it corresponds to the highest peak on the "Accuracy vs K Value" resulting in maximum test accuracy of
0.954. This value represents the optimal balance between:
Low K (e.g., K=1): Which is highly sensitive to noise and prone to overfitting.
High K (e.g., K=60): Which oversmoothes the decision boundary and can lead to underfitting.
````

Reinforcement Learning is essentially how we teach a computer program to learn by trial and error, much like training a dog or teaching a child to ride a bike. In this system, called an Agent, the program takes an action in an Environment (like a video game or a real-world simulation). If the action is good (like scoring a point or balancing the bike), the agent gets a Reward (a positive numerical score). If the action is bad (like crashing or losing a point), it gets a Penalty (a negative score). The agent's goal is always to maximize the total reward it receives over time, continuously adjusting its strategy based on the feedback it gets.
This method is powerful because the agent learns the best strategy without being explicitly told what to do; it figures it out itself. For example, to teach a computer to play chess, you don't program every single move. Instead, you reward the agent for winning and penalize it for losing, and it gradually develops sophisticated, winning tactics. RL is the technology behind self-driving cars, robotic control, and is famous for creating AI that can beat human champions at complex games like Go and Dota 2.

Oversampling and Undersampling are techniques used to deal with imbalanced datasets, which happen when one category (or class) has far fewer examples than the other. For instance, if you have 10,000 credit card transactions and only 100 are fraudulent, your data is highly imbalanced. A machine learning model trained on this skewed data might ignore the small class, leading to poor performance.

To fix this, we balance the classes. Oversampling means creating new, synthetic examples of the minority class to match the size of the majority class (e.g., using techniques like SMOTE). Undersampling means randomly deleting examples from the majority class until it matches the size of the minority class. While both aim to level the playing field, Undersampling can lose valuable information from the deleted examples, while Oversampling can sometimes lead to overfitting.
NOTE:The choice between them depends on the specific dataset and problem.
